## 🧹 Lesson 1: Data Cleaning (and its Mathematical Heart)

### 🎯 Intuitive Idea
Data cleaning is the process of making your datasets reliable an ready for the analysis. Mathematically, think of it like removing noise so your function can learn real relationship, not random error.

### Why it matters
Model like linear regression, logistic regression, or neural networks all assume that your data:
1. is numerical and normalized,
2. has no outliers dominating the loss function,
3. reflects true process, not random recording error

Bad data → bad gradient directions → poor optimization → wrong predictions.

```python
import pandas as pd
import numpy as np

# Simulate messy data
data = {
    'Age': [25, 30, np.nan, 22, 28, 120, 27],
    'Salary': [50000, 54000, 58000, np.nan, 62000, 600000, 61000],
    'Department': ['Sales', 'HR', 'IT', 'Sales', None, 'HR', 'IT']
}

df = pd.DataFrame(data)
print("Raw Data:\n", df)

# 1️⃣ Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Department'] = df['Department'].fillna('Unknown')

# 2️⃣ Detect and cap outliers using z-score
z_scores = np.abs((df['Salary'] - df['Salary'].mean()) / df['Salary'].std())
df.loc[z_scores > 2.5, 'Salary'] = df['Salary'].median()

# 3️⃣ Standardize numeric features
df['Age'] = (df['Age'] - df['Age'].mean()) / df['Age'].std()
df['Salary'] = (df['Salary'] - df['Salary'].mean()) / df['Salary'].std()

print("\nCleaned & Scaled Data:\n", df)
```
